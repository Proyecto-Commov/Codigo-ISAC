# Pruebas sobre los canales $H[m,k]$

```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
from channel_estimator import *
from channel_features import *
```

```python
base_dir = Path.cwd().resolve()
repo_root = None
for candidate in [base_dir, *base_dir.parents]:
    if (candidate / "data").exists():
        repo_root = candidate
        break

if repo_root is None:
    raise FileNotFoundError("No se pudo localizar la raíz del repositorio a partir del directorio actual")

yaml_path = repo_root / "data" / "Modulator.yaml"

tx_path = repo_root / "data" / "vacio" / "iq_tx.bin"
rx_path = repo_root / "data" / "vacio" / "iq_rx.bin"

print(f"Raíz detectada: {repo_root}")
print(f"YAML: {yaml_path}")
print(f"TX: {tx_path}")
print(f"RX: {rx_path}")
```

## Vacío

```python
H = compute_channel_matrix_from_iq_paths(
    tx_path,
    rx_path,
    yaml_path,
    n_symbols=100,
    storage_format_tx="complex64",
    storage_format_rx="complex64",
    fftshift=False,
    output_order="mk",
    verbose=False,
)

features = extract_channel_matrix_features(
    H,
    input_order="mk",
    percentiles=(25, 50, 75),
)

print_channel_matrix_features(features)
```
